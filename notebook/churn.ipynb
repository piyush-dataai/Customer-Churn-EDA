```python
def conv(value):
    if value ==1:
        return "yes"
    else:
        return "no"
df["SeniorCitizen"] = df["SeniorCitizen"].apply(conv)
```

```python
sns.countplot(x="Churn", hue="Churn", data=df, palette="coolwarm", legend=False)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["tenure"], bins=30, kde=True, color="skyblue")

plt.title("Distribution of Customer Tenure", fontsize=14)
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Churn",
    y="MonthlyCharges",
    hue="Churn",
    data=df,
    palette="coolwarm",
    legend=False
)

plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
contract_churn = pd.crosstab(df["Contract"], df["Churn"], normalize="index") * 100

ax = contract_churn.plot(
    kind="bar",
    stacked=True,
    colormap="coolwarm",
    figsize=(10,6)
)

plt.title("Churn Percentage by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", label_type="center")

plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# --- 🎨 COLOR THEORY THEME SETUP ---
# Background: Muted Slate (#f0f4f7) - reduces eye strain
# Accent Blue: Professional Navy (#2c3e50) - for text/titles
# Grid: Clean White (#ffffff)
bg_color = "#f0f4f7"
text_color = "#2c3e50"

sns.set_style("whitegrid", {
    "axes.facecolor": bg_color,
    "grid.color": "#ffffff",
    "patch.edgecolor": "none"
})

# Create the figure with the background color
fig = plt.figure(figsize=(24, 16), facecolor=bg_color)

# 🔥 Dashboard Title - Positioned for better white space
plt.suptitle("CUSTOMER CHURN ANALYSIS DASHBOARD", 
             fontsize=28, fontweight="bold", color=text_color, y=0.96)

# 1. Churn Count - Using a clean palette
plt.subplot(2, 3, 1)
ax1 = sns.countplot(x="Churn", hue="Churn", data=df, palette="coolwarm", legend=False)
plt.title("Customer Churn Distribution", fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Churn Status", fontweight="bold")
plt.ylabel("Total Count", fontweight="bold")

# 2. Tenure Histogram - Changed color to match the 'coolwarm' blue for consistency
plt.subplot(2, 3, 2)
sns.histplot(df["tenure"], bins=30, kde=True, color="#4a73b2", edgecolor="white")
plt.title("Distribution of Customer Tenure", fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Tenure (Months)", fontweight="bold")
plt.ylabel("Number of Customers", fontweight="bold")

# 3. Monthly Charges vs Churn - Boxplot
plt.subplot(2, 3, 3)
sns.boxplot(x="Churn", y="MonthlyCharges", hue="Churn", 
            data=df, palette="coolwarm", legend=False, width=0.6)
plt.title("Monthly Charges vs Churn", fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Churn", fontweight="bold")
plt.ylabel("Monthly Charges ($)", fontweight="bold")

# 4. Contract vs Churn (%) - Stacked Bar
plt.subplot(2, 3, 4)
contract_churn = pd.crosstab(df["Contract"], df["Churn"], normalize="index") * 100
ax4 = contract_churn.plot(kind="bar", stacked=True, colormap="coolwarm", ax=plt.gca(), width=0.7)

plt.title("Churn % by Contract Type", fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Contract Type", fontweight="bold")
plt.ylabel("Percentage", fontweight="bold")

# Bold percentage labels for readability against dark backgrounds
for container in ax4.containers:
    ax4.bar_label(container, fmt="%.1f%%", label_type="center", 
                  fontsize=11, fontweight="bold", color="white")

plt.legend(title="Churn", bbox_to_anchor=(1.0, 1.0), loc='upper left')

# 5. Correlation Heatmap - Better sizing and color contrast
plt.subplot(2, 3, 5)
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0,
            fmt=".2f", annot_kws={"weight": "bold", "size": 11},
            cbar_kws={'shrink': .8})
plt.title("Feature Correlation Heatmap", fontsize=15, fontweight="bold", pad=15)

# Finalizing layout
plt.tight_layout(rect=[0, 0.03, 1, 0.93]) 

# Adding a small footer for extra professional touch
fig.text(0.99, 0.01, 'Data Analysis Dashboard | 2026', 
         horizontalalignment='right', fontsize=10, color='gray', style='italic')

plt.show()
```
